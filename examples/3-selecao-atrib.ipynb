## Seleção de atributos (Feature Selection)

- O que faz: Escolhe um subconjunto das variáveis originais mais relevantes.
- Mantém a interpretabilidade: Ao contrário do PCA. os atributos escolhidos continuam sendo os mesmos do conjunto original (ex: IDADE, SALDO_ATUAL, etc).
- Objetivo: Eliminar atributos irrelevantes ou redundantes para melhorar o desempenho e simplicidade do modelo.

> Dados usados: [ad.data](../anexos/ad.data)

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score
from sklearn.feature_selection import chi2, SelectKBest
# Onde chi2 (chi-squared) é a técnica para medir a dependencia dos atributos com o alvo 
# e
# SelectKBest é o método de seleção dos K melhores atributos calculados por chi2

anuncio = pd.read_csv('../anexos/ad.data', header=None)
anuncio.shape # 3279 linhas e 1559 colunas
```

### Separando os valores X de Y

```python
X = anuncio.iloc[:,:-1].values
y = anuncio.iloc[:,-1].values

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)
```

### Criando e Prevendo o Modelo

```python
modelo1 = GaussianNB()
modelo1.fit(X_train, y_train)

previsoes1 = modelo1.predict(X_test)
print(accuracy_score(y_test, previsoes1))
```

### Aplicando a Seleção de Atributos

```python
selecao = SelectKBest(chi2, k=7)
X_novo = selecao.fit_transform(X, y) # Se passa o y para a tecnica manter a interpretabilidade entre os atributos e o alvo

X_novo
```

```python
X_novo.shape # agora são 3279 linhas e apenas 7 colunas
```

### Separando os novos valores de X

```python
X = anuncio.iloc[:,:-1].values
y = anuncio.iloc[:,-1].values

X_train, X_test, y_train, y_test = train_test_split(X_novo, y, test_size=0.3, random_state=0)
```

### Criando o novo modelo

```python
modelo2 = GaussianNB()
modelo2.fit(X_train, y_train)

previsoes2 = modelo2.predict(X_test)
print(accuracy_score(y_test, previsoes2))
```
